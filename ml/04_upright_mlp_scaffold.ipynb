# 04. Upright Posture MLP Architecture Scaffold (Week 3 Preparation)

> **Notice:** In Week 2, posture detection uses a heuristic upright score (`signals.upright = 1.0`).
> This notebook prepares the architecture for Week 3 where 33 full-body MediaPipe landmarks
> and phone accelerometer vectors feed an MLP classifier to distinguish lying down vs upright standing/sitting.

```python
import tensorflow as tf
from tensorflow.keras import layers, models

def build_upright_mlp_scaffold(input_dim=99) -> tf.keras.Model:
    """
    Inputs: 33 3D body landmarks (x, y, z) = 99 features
    Output: Single scalar probability (0=lying in bed, 1=upright standing/sitting)
    """
    inputs = layers.Input(shape=(input_dim,), name="pose_landmarks_3d")
    x = layers.Dense(128, activation="relu")(inputs)
    x = layers.Dropout(0.2)(x)
    x = layers.Dense(64, activation="relu")(x)
    x = layers.Dense(32, activation="relu")(x)
    outputs = layers.Dense(1, activation="sigmoid", name="is_upright")(x)

    model = tf.keras.Model(inputs=inputs, outputs=outputs, name="Upright_Posture_MLP_Week3")
    return model

scaffold = build_upright_mlp_scaffold()
scaffold.summary()
print("Week 3 upright MLP scaffold compiled successfully.")
```
